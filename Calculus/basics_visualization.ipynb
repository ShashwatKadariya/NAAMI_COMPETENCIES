```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
# --- Convex vs Non-Convex Functions ---
x = np.linspace(-3, 3, 100)
y_convex = x**2
y_non_convex = x**3 - 3*x
```

```python
plt.figure(figsize=(12, 6))
```

```python
plt.subplot(1, 2, 1)
plt.plot(x, y_convex, label='$f(x) = x^2$', color='b')
plt.title("Convex Function: $f(x) = x^2$")
plt.axhline(0, color='black',linewidth=1)
plt.axvline(0, color='black',linewidth=1)
plt.grid(True)
plt.legend()
```

```python
plt.subplot(1, 2, 2)
plt.plot(x, y_non_convex, label='$f(x) = x^3 - 3x$', color='r')
plt.title("Non-Convex Function: $f(x) = x^3 - 3x$")
plt.axhline(0, color='black',linewidth=1)
plt.axvline(0, color='black',linewidth=1)
plt.grid(True)
plt.legend()
```

```python
# --- Derivative Properties ---
```

```python
def f(x):
    return x**2

def f_prime(x):
    return 2*x
```

```python
x_vals = np.linspace(-3, 3, 100)
y_vals = f(x_vals)
y_prime_vals = f_prime(x_vals)
```

```python
plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals, label='$f(x) = x^2$', color='b')
plt.plot(x_vals, y_prime_vals, label="$f'(x) = 2x$", color='r', linestyle='--')
plt.title("Function and its Derivative")
plt.axhline(0, color='black',linewidth=1)
plt.axvline(0, color='black',linewidth=1)
plt.grid(True)
plt.legend()
```

```python
# --- Chain Rule Example ---
# Function f(g(x)) = sin(x^2)
```

```python
def g(x):
    return x**2

def f(g):
    return np.sin(g)

def chain_rule(x):
    return np.cos(x**2) * 2*x
```

```python
x_vals = np.linspace(-3, 3, 100)
y_chain = f(g(x_vals))
y_chain_prime = chain_rule(x_vals)
```

```python
plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_chain, label='$f(g(x)) = sin(x^2)$', color='b')
plt.plot(x_vals, y_chain_prime, label="Derivative using chain rule", color='r', linestyle='--')
plt.title("Chain Rule Example: $f(g(x)) = sin(x^2)$")
plt.axhline(0, color='black',linewidth=1)
plt.axvline(0, color='black',linewidth=1)
plt.grid(True)
plt.legend()
```

```python
# --- Partial Derivatives ---
# f(x, y) = x^2 + y^2
```

```python
def f_partial(x, y):
    return x**2 + y**2
def partial_x(x, y):
    return 2*x
def partial_y(x, y):
    return 2*y
```

```python
x_vals = np.linspace(-3, 3, 50)
y_vals = np.linspace(-3, 3, 50)
X, Y = np.meshgrid(x_vals, y_vals)

Z = f_partial(X, Y)
Zx = partial_x(X, Y)
Zy = partial_y(X, Y)
```

```python
fig = plt.figure(figsize=(18, 5))
```

```python
fig = plt.figure(figsize=(18, 5))

ax1 = fig.add_subplot(1, 3, 1, projection='3d')
ax1.plot_surface(X, Y, Z, cmap='viridis')
ax1.set_title('Function $f(x, y) = x^2 + y^2$')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_zlabel('f(x, y)')

ax2 = fig.add_subplot(1, 3, 2, projection='3d')
ax2.plot_surface(X, Y, Zx, cmap='plasma')
ax2.set_title('Partial Derivative $\\frac{∂f}{∂x}$')
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_zlabel('∂f/∂x')

ax3 = fig.add_subplot(1, 3, 3, projection='3d')
ax3.plot_surface(X, Y, Zy, cmap='cividis')
ax3.set_title('Partial Derivative $\\frac{∂f}{∂y}$')
ax3.set_xlabel('x')
ax3.set_ylabel('y')
ax3.set_zlabel('∂f/∂y')

plt.tight_layout()
plt.show()
```

```python
# --- Jacobian Matrix ---
#  F(x, y) = [f1(x, y), f2(x, y)] = [x^2 + y^2, x * y]
```

```python
def f1(x, y):
    return x**2 + y**2

def f2(x, y):
    return x * y

def jacobian(x, y):
    df1_dx = 2*x
    df1_dy = 2*y
    df2_dx = y
    df2_dy = x
    return np.array([[df1_dx, df1_dy], [df2_dx, df2_dy]])
```

```python
jacobian_matrix = jacobian(1, 2)
print("Jacobian Matrix at (x=1, y=2):")
print(jacobian_matrix)
```

```python
x_vals = np.linspace(-3, 3, 100)
y_vals = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x_vals, y_vals)
F1 = f1(X, Y)
F2 = f2(X, Y)
```

```python
plt.figure(figsize=(8, 6))
plt.quiver(X, Y, F1, F2, angles='xy', scale_units='xy', scale=1)
plt.title("Vector Field for $f_1(x, y) = x^2 + y^2$ and $f_2(x, y) = x*y$")
plt.xlim([-3, 3])
plt.ylim([-3, 3])
plt.axhline(0, color='black',linewidth=1)
plt.axvline(0, color='black',linewidth=1)
plt.grid(True)
plt.show()
```
